```python
# The collatz game.
def collatz(number):
    if int(number) % 2 == 0:
        result = int(number) // 2
        print(result, end = " ")
        return result
    else:
        #print(number, end = " ")
        result = (3 * int(number)) + 1
        print(result, end = " ")
        return result
        
print("Collatz Code Program")

while True:
    try:
        user_num = int(input())
        print(user_num, end = " ")
        break
    except ValueError:
        print("You have to enter an integer.")

while int(user_num) != 1:
    user_num = collatz(user_num)
```

```python
# The box printer code.
print("Welcome to the box printer.")
user_symbol = str(input("Enter the symbol you want: "))
user_width = int(input("Enter your preferred width: "))
user_height = int(input("Enter your preferred height: "))

def box_print(user_symbol, user_width, user_height):
    if len(user_symbol) != 1:
      raise Exception("Symbol must be a single character string.")
    if user_width <= 2:
      raise Exception("Width must be greater than 2.")
    if user_height <= 2:
      raise Exception("Height must be greater than 2.")

    print(user_symbol * user_width)
    for i in range(user_height - 2):
        print(user_symbol + (' ' * (user_width - 2)) + user_symbol)
    print(user_symbol * user_width)

try:
    box_print(user_symbol, user_width, user_height)
except Exception as err:
    print("An exception happened: " + str(err))
```

```python
# Other Double Underscore Methods include:
class Arithmetics:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the addition (+) operator
    def __add__(self, other):
        # Always return a brand new instance of your class
        return Arithmetics(self.x + other.x, self.y + other.y)

    # Overloading the subtraction (-) operator
    def __sub__(self, other):
        return Arithmetics(self.x - other.x, self.y - other.y)

    # Overloading the multiplication (*) operator
    def __mul__(self, other):
        # Can handle multiplying by a scalar number (e.g., Vector * 3)
        if isinstance(other, (int, float)):
            return Arithmetics(self.x * other, self.y * other)
        # Or handle vector dot product multiplication (Vector * Vector)
        return self.x * other.x + self.y * other.y

    # User friendly display format
    def __str__(self):
        return f"Arithmetics({self.x}, {self.y})"

a1 = Arithmetics(2, 4)
a2 = Arithmetics(5, 1)

# 1. Triggers a1.__add__(a2)
a3 = a1 + a2
print(a3)

# 2. Triggers a1.__sub__(a2)
a4 = a1 - a2
print(a4)

# 3. Triggers a1.__mul__(3) scalar multiplication
a5 = a1 * 3
print(a5)

# Testing
a6 = a1 * a2
print(a6)
```

```python
# Stacks and Queues in Object Oriented Programming:
# Stacks use the Last-In, First-Out principle where the last item appended to the list is the first item to come out using append and pop.
# LI-FO
class CardboardCup:
    def __init__(self, owner):
        self.owner = owner

class CupStack:
    def __init__(self):
        # The internal list used to hold our cups
        self._cups = []
        
    def push(self, owner, cup):
        self._cups.append(cup)  # Push to the top of the stack
        
    def pop(self):
        # Safety check: do not pop if the stack is empty!
        if self.is_empty():
            print("Action Blocked: The stack is empty!")
            return None
        return self._cups.pop()  # Pop from the top of the stack
        
    def is_empty(self):
        return len(self._cups) == 0

stack = CupStack()
cup_1 = CardboardCup("Alex")
cup_2 = CardboardCup("Belle")

stack.push(cup_1, "Latte")
stack.push(cup_2, "Coffee")

print(stack.pop())
print(stack.pop())
print(stack.is_empty())

print("~" * 10)
# Queues work with the First-In, First-Out principle using the same append but with pop(0) in order to extract the first item in the list.
# FI-FO
class CupQueue:
    def __init__(self):
        self._cups = []
        
    def enqueue(self, owner, cup):
        self._cups.append(cup)  # Add to the back of the line
        
    def dequeue(self):
        if self.is_empty():
            print("Action Blocked: The conveyor chute is empty!")
            return None
        return self._cups.pop(0)  # Remove from the front of the line (Index 0)
        
    def is_empty(self):
        return len(self._cups) == 0

queue = CupQueue()
queue.enqueue(cup_1, "Milo")
queue.enqueue(cup_2, "Lipton")

print(queue.dequeue())
print(queue.dequeue())
print(queue.is_empty())
```

```python
# Another use case of stacks and queues is the deque approach
from collections import deque

# ==========================================
# 1. THE PERFECT QUEUE (FIFO: First In, First Out)
# ==========================================
queue = deque()

# Pushing elements to the back
queue.append("Customer 1")
queue.append("Customer 2")

# Removing elements from the front - Strict O(1)!
# Lists using .pop(0) would take slow O(n) time here.
first_up = queue.popleft() 
print(f"Served: {first_up}")  # Outputs: Served: Customer 1


# ==========================================
# 2. THE HIGH-PERFORMANCE STACK (LIFO: Last In, First Out)
# ==========================================
stack = deque()

# Pushing elements to the top
stack.append("Page 1")
stack.append("Page 2")

# Popping elements from the top - Strict O(1)
top_page = stack.pop()
print(f"Closed: {top_page}")  # Outputs: Closed: Page 2
```

```python
# Binary Search Tree Node (BST Node):
class CupNode:
    def __init__(self, price):
        self.price = price  # The key value used for sorting
        self.left = None    # Pointer to smaller cup node (cheaper)
        self.right = None   # Pointer to larger cup node (more expensive)


# Standalone recursive insertion function:
def insert(node, price):
    # Base Case: If we reach an empty spot, stamp out a new node here!
    if node is None:
        return CupNode(price)
        
    # Recursive Case: Decide which branch to follow
    if price < node.price:
        # If the new price is cheaper, go left
        node.left = insert(node.left, price)
    elif price > node.price:
        # If the new price is more expensive, go right
        node.right = insert(node.right, price)
        
    return node  # Return the unchanged node pointer


# --- Running the code ---
root = CupNode(4.00)
root = insert(root, 3.00)
root = insert(root, 5.00)

print(root.left.price)   # Outputs: 3.0
print(root.right.price)  # Outputs: 5.0

# When you want to search the BST you have to write an independent recursive search function that helps compares the target with the 
# current node in the tree, if the target is larger it checks the right branch of the tree and if it's smaller then it checks the
# left branch instead.
def search(node, target): # The Search Recursion Function...
    # Base Cases: we reached the end of the tree, or we found the match!
    if node is None or node.price == target:
        return node
        
    # Recursive Cases: branch left or right
    if target < node.price:
        return search(node.left, target)
    else:
        return search(node.right, target)

result_node = search(root, 4.00)

print("~" * 20)
if result_node is not None:
    print(f"The target price was found and here is the output: {result_node.price}")
else:
    print("Nothing was found")
```

```python
# Sorting in Python (Bubble, Insertion and Selection Sorting):
# Basic Sorting in Python using Tuple Sorting:
flavors = ["Mocha", "Latte"]

# The Tuple Swap: Python evaluates both on the right first, then assigns them to the left (a,b = b,a)
flavors[0], flavors[1] = flavors[1], flavors[0]

print(flavors) # Output: ['Latte', 'Mocha'] (Safely swapped!)

# Bubble Sorting:
# Don't forget that the range method allows you to include the first item in your range but doesn't allow you to include the last item in
# the list. So range(0, 4) allows you to show 0, 1, 2, and 3 not 4.
def bubble_sort(arr):
    n = len(arr)
    # When bubble sorting, you use nested loops where the first loop / outer loop determines how many times the inner loop
    # must repeat the "bubbling" process in order to guarantee the entire list is sorted while the second loop compares the items in the
    # list and ensures that at each iteration the largest item in the list is moved to the end of the list.
    for pass_num in range(n):
        # Implementing the pass_num as -pass_num in the end of the range is a way of reducing repetition in the sorting process, in the
        # sense that when bubble sorting, each loops iteration moves the largest item to the end of the list and progressively moves
        # leftwards, so the way it works is when pass_num starts at 0 and assuming the length of the list is 4 then the calculation
        # becomes 4 - 0 - 1 which will equal to 3 and that 3 index would be excluded meaning the loop will run for 0, 1 and 2, then 
        # when pass_num becomes 1 the calculation becomes 4-1-1 which is equal to 2 and the interation becomes 0,1 and so on. This allows
        # us to remoove redundant repetition of comparing the last items in the list even though we already know that the last items are
        # properly sorted.
        for i in range(0, n - pass_num - 1):
            if arr[i] > arr[i + 1]:
                # Swap the adjacent cups!
                arr[i], arr[i + 1] = arr[i + 1], arr[i]

prices = [4.50, 3.50, 5.00, 2.00]
bubble_sort(prices)
print(prices)

# Selection Sorting:
# Selection Sorting only moves the smallest item in the list to the beginning of the list, as compared to bubble sorting that sorts from
# the end of the list, selection sorting works from the start of the list.
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i  # Assume the current item is the smallest
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j  # Update the index of the smallest item found
                
        # Swap the smallest item with the first unsorted item
        arr[i], arr[min_index] = arr[min_index], arr[i]

pricez = [4.50, 3.50, 5.00, 2.00]
selection_sort(pricez)
print(pricez)

# Insertion Sorting:
# Insertion Sorting works in a way that it seems like it's arranging the list from the middle of the list instead of from an end like the
# Bubble Sorting or the Selection Sorting functions, in another way it also sorts items from the left to the right. For each item, it picks
# it up and slides it backward (left) through the sorted section until it finds its correct relative position.
def insertion_sort(arr):
    # The first loop goes through the list and allos this selection method to run enough times for the entire list to be sorted.
    for i in range(1, len(arr)):
        # We start with creating a temporary variable copy of the position of i in the list.
        key = arr[i]  # Pick up the current cup
        # Then we assign j to be the index of the item to the left of i or key
        j = i - 1
        
        # This loop slides the cup left as long as the neighboring cup is more expensive
        while j >= 0 and arr[j] > key:
            # This line copies the item at j to the item on the right for instance, [20,30,15] becomes [20,30,30] when i is 2 and j is 1
            arr[j + 1] = arr[j]  # Shift the more expensive cup to the right
            # This changes the value of j to be a step lower so that the list is sorted in ascending order for instance [20,30,30] turns to
            # [20,20,30] when j changes to 0 and j+1 becomes 1
            j -= 1
        # This helps us to replace the very first index with the value of the smallest item which has been preserved in key all this while
        # for instance after j == 0, then j turns to -1 before the while loop stops, so the value of j+1 which is -1+1 = 0, the first index
        # turns to the value of key.
        arr[j + 1] = key  # Drop the cup into its correct spot

pricest = [4.50, 3.50, 5.00, 2.00]
insertion_sort(pricest)
print(pricest)
```

```python
# Efficient Sorting Methods:
# Merge Sorting:
def merge_sort(arr):
    # BASE CASE: A list of 0 or 1 elements is already sorted!
    if len(arr) <= 1:
        return arr

    # 1. DIVIDE: Find the midpoint and split the array
    mid = len(arr) // 2 # double slashes and single slash represent division however double returns the whole number while single returns
    # a float number. Take note that it doesn't approximate the value
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Recursive calls to break down the left and right halves
    merge_sort(left_half)
    merge_sort(right_half)

    # 2. CONQUER/MERGE: Combine the sorted halves back into 'arr'
    i = 0  # Pointer for left_half
    j = 0  # Pointer for right_half
    k = 0  # Pointer for the main arr
    # All elements start from index/number 0
    
    # Compare elements from both halves and place the smaller one into arr
    while i < len(left_half) and j < len(right_half):
        if left_half[i] < right_half[j]:
            arr[k] = left_half[i]
            i += 1
        else:
            arr[k] = right_half[j]
            j += 1
        k += 1

    # If there are any leftover elements in left_half, copy them over
    while i < len(left_half):
        arr[k] = left_half[i]
        i += 1
        k += 1

    # If there are any leftover elements in right_half, copy them over
    while j < len(right_half):
        arr[k] = right_half[j]
        j += 1
        k += 1

    return arr

# OR
# This method seperates the two parts of the merge sorting method into two separate functions.
# The first function splits the list into two halves and calls the second function to create a new organized list after comparing all
# elements in both lists.
def merge_sort(arr):
    # 1. BASE CASE
    if len(arr) <= 1:
        return arr
        
    # 2. THE SPLIT (Divide)
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])   # Sort the left side recursively
    right_half = merge_sort(arr[mid:])  # Sort the right side recursively
    
    # 3. THE COMBINE (Conquer)
    return merge(left_half, right_half) # Merge the two sorted halves

# The second function creates an empty list and sort the splitted lists in the correct order by comparing each item against each item in
# both lists, and then appends the result into the empty list and extends the empty list to accommodate the remnants.
def merge(left, right):
    sorted_list = []
    i = 0  # Pointer for the left list
    j = 0  # Pointer for the right list
    
    # Compare the front elements of both lists and copy the smaller one
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1
            
    # Copy any remaining items from both lists
    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
    
    return sorted_list

# Quick Sorting:
# This is a sorting method that selects the mid item in the list (pivot) and moves items smaller than the pivot to the left while it moves
# items larger than the picot to the right, then it returns a recursion of itself so that it sorts all items in the left and rights lists
# properly.
def quicksort(arr):
    # 1. BASE CASE
    if len(arr) <= 1:
        return arr
        
    # 2. SELECT PIVOT (We choose the middle cup as the pivot)
    pivot = arr[len(arr) // 2]
    
    # 3. PARTITION (Divide)
    # Using List Comprehension....
    left = [x for x in arr if x < pivot]    # All cups cheaper than the pivot
    middle = [x for x in arr if x == pivot] # All cups equal to the pivot
    right = [x for x in arr if x > pivot]   # All cups more expensive than the pivot
    
    # 4. RECURSIVE CONQUER (Combine)
    return quicksort(left) + middle + quicksort(right)
```

```python
def draw_symbol(char, height=7, width=5, fill="X", space=" "):
    char = char.upper()
    if height < 3 or width < 3:
        print("Height must be >= 3 and Width >= 3.")
        return

    mid_r = height // 2
    mid_c = width // 2

    for r in range(height):
        row_str = ""
        for c in range(width):
            # Positional boolean checks
            is_top = (r == 0)
            is_bot = (r == height - 1)
            is_mid_r = (r == mid_r)
            is_left = (c == 0)
            is_right = (c == width - 1)
            is_mid_c = (c == mid_c)

            # Diagonals for letters like X, Z, V, Slash
            is_diag_main = (r * width == c * height)
            is_diag_anti = ((r + 1) * width == (width - c) * height)

            # Letter geometry definitions
            draw = False

            if char == 'A':
                draw = (is_top and not is_left and not is_right) or \
                       (is_left and not is_top) or \
                       (is_right and not is_top) or \
                       is_mid_r

            elif char == 'B':
                draw = is_left or \
                       (is_top or is_bot or is_mid_r) and not is_right or \
                       (is_right and not is_top and not is_bot and not is_mid_r)

            elif char == 'C':
                draw = (is_left and not is_top and not is_bot) or \
                       ((is_top or is_bot) and not is_left)

            elif char == 'D':
                draw = is_left or \
                       ((is_top or is_bot) and not is_right) or \
                       (is_right and not is_top and not is_bot)

            elif char == 'E':
                draw = is_left or is_top or is_mid_r or is_bot

            elif char == 'F':
                draw = is_left or is_top or is_mid_r

            elif char == 'H':
                draw = is_left or is_right or is_mid_r

            elif char == 'I':
                draw = is_top or is_bot or is_mid_c

            elif char == 'L':
                draw = is_left or is_bot

            elif char == 'O':
                draw = ((is_left or is_right) and not is_top and not is_bot) or \
                       ((is_top or is_bot) and not is_left and not is_right)

            elif char == 'T':
                draw = is_top or is_mid_c

            elif char == 'U':
                draw = ((is_left or is_right) and not is_bot) or \
                       (is_bot and not is_left and not is_right)

            elif char == 'Z':
                draw = is_top or is_bot or (c == width - 1 - int(r * (width / height)))

            row_str += fill if draw else space
        print(row_str)

# Examples
print("--- Letter A (7x5) ---")
draw_symbol('A', height=7, width=5)

print("\n--- Letter O (7x5) ---")
draw_symbol('O', height=7, width=5)

print("\n--- Letter H (9x7) ---")
draw_symbol('H', height=9, width=7)
```

```python
def string_to_array(s):
    if s == "":
        return [""]
    
    result = []
    word = ""
    for char in s:
        if char != " ":
            word = word + char
        else:
            result.append(word)
            word = ""
    
    if word:
        result.append(word)
    
    return result
```

```python
leads = [
    "   Sarah Johnson | SARAH@EMAIL.COM | 5000 ",
    "michael brown | michael@email.com | 2500 ",
    " AMANDA SMITH | amanda@email.com | 12000 ",
    "john doe | JOHN@EMAIL.COM | 800 ",
]

def clean_leads(evidence):
    cleaned_leads = [{}, {}, {}, {}]
    for index,lead in enumerate(evidence):
        name, email, budget = lead.strip().split(" | ")

        cleaned_leads[index]["name"] = name.title()
        cleaned_leads[index]["email"] = email.lower()
        cleaned_leads[index]["budget"] = int(budget)
        
    return cleaned_leads

cleaned_data = clean_leads(leads)
print(cleaned_data)

def clean_leadz(data):
    cleaned_leads = []
    for lead in data:
        name, email, budget = lead.strip().split(" | ")
        cleaned_leads.append({"name": name.title(), "email": email.lower(), "budget": int(budget)})
        
    return cleaned_leads

cleaned_datas = clean_leadz(leads)
print(cleaned_datas)
```

```python
# Parsing simply means the act of converting a string data type into a number data type for calculations.
test = "     167267376          "

print(int(test))

# def parse_nigerian_number(input_string):
def parse_nigerian_number(input_string):
    """Convert a Nigerian-style number string to a float."""
    # Remove commas (used as thousand separators)
    cleaned = input_string.replace(",", "")
    # Remove currency symbols
    cleaned = cleaned.replace("₦", "")
    # Remove extra whitespace
    cleaned = cleaned.strip()
    # Convert to float
    try:
        return float(cleaned)
    except ValueError:
        # Try to handle other formatting issues
        cleaned = cleaned.replace(" ", "")
        return float(cleaned)

user_input = input("Enter your amount:")
print(parse_nigerian_number(user_input))
# print(parse_nigerian_number("₦150,000.00"))
```

```python
class KitchenCalculator:
    def calculate_inventory(self):
        customers = 3
        cereal_per_customer = 2.5
        milk_per_customer = 1.25
        
        starting_cereal = 10.0
        starting_milk = 15.0
        
        # 1. Use multiplication to calculate the totals needed
        total_cereal = cereal_per_customer * customers
        total_milk = milk_per_customer * customers
        
        # 2. Use subtraction to calculate the remaining inventory
        remaining_cereal = starting_cereal - total_cereal
        remaining_milk = starting_milk - total_milk
        
        return remaining_cereal, remaining_milk

test = KitchenCalculator()

print(test.calculate_inventory())
```

```python
# Dictionaries:
names = {"Dami": 12, "Drake": 15, "Fiona": 30, "Flora": 45}

for name in names:
    print(name) # This prints the keys
    print(names[name]) # This prints out the values of the keys in the dictionary.

names["new_name"] = 50 # This is another way to add a new key and value to a dictionary asided update method
print(names)

print("~" * 20)

# Dictionary Methods:
# GET: The syntax of ".get" for dictionaries is dict.get(key, default = None)
print(names.get("Dami")) # This extracts the value of a known key, if the key doesn't exists, output = 0
print(names.get("Flora", 10)) # This gets the value of a known key and puts a default incase it doesn't exist
print(names.get("Rume", 10)) # This prints the default value specified since the key doesn't exist.
print(names) # Using .get doesn't add or change the original dictionary

# UPDATE: The syntax of ".update" for dictionaries is dict.update(other_dict_or_iterable)
print(names.update({"Deone": 10})) # Using print with update method would give None.
print(names) # Pay attention to the two brackets used for the update method.
names.update({"Dami": 30}) # Update method also modifies the values assigned to the keys in a dictionary
print(names)
print(names.update({"Rume": 23}))
print(names)

# POP: The syntax of ".pop" for dictionaries is "dict.pop(key, default)"
print(names.pop("Flora")) # This removes the value of the known key you specified and prints it
try:
    print(names.pop("Clara")) # This code would cause an error on python because this key down't exist in the dictionary
except:
    print("This key is not in the dictionary")
print(names.pop("Clara", 15)) # This code prints the default value if the key you specified doesn't exist in the dictionary

# OTHER METHODS IN DICTIONARIES:
# KEYS: The syntax of ".keys" for dictionaries is "dict.keys()"
print(names.keys()) # This is used to get all the keys in the dictionary

# VALUES: The syntax of ".values" in dictionaries is "dict.values()"
print(names.values()) # This is used to get all the values in the dictionary

# ITEMS: The syntax of ".items" in dictionaries is "dict.items()"
print(names.items()) # This is used to get all the items in the dictionary including the keys and the values

#SETDEFAULT: The syntax of ".setdefault" in dictiopnaries is "dict.setdefault(key, default=None)"
print(names.setdefault("Favour", None)) # This method checks the acts like GET and returns the default when the key doesn't exist in the dictionary

# POPITEM: The syntax of ".popitem" in dictionaries is "dict.popitems()"
print(names.popitem()) # This method removes the last item in the dictionary

# COPY: The syntax of ".copy" in dictionaries is "dict.copy()"
name_copy = names.copy() # This copies the keys and values in the dictionary
print(name_copy)

# CLEAR: The syntax of ".clear" in dictionaries is "dict.clear()"
print(names.clear()) # This method removes every key and value in the dictionary

# FROMKEYS: The syntax of ".fromkeys()" in dictionaries is "dict.fromkeys(iterable, value=None)"
print(name_copy.fromkeys(['Dami', 'Rume'], 0)) # This method creates a new dictionary with the keys you specify and the values you input.
```
