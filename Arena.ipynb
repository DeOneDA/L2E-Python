```python
def safe_calculator(a, operator, b):
    if operator == "/" and b == 0:
        return "Cannot divide by zero"
    elif operator == "%" and b == 0:
        return "Cannot divide by zero"
    elif operator == "+":
        result = a + b
        return result
    elif operator == "-":
        result = a - b
        return result
    elif operator == "*":
        result = a * b
        return result
    elif operator == "/":
        result = float(f"{a/b:.2f}")
        return result
    elif operator == "%":
        result = a % b
        return result
    elif operator == "**":
        result = a ** b
        return result
    else:
        return "Invalid operator"
```

```python
def exact_calculator(left, operator, right):
    try:
        a = int(left)
        b = int(right)
    except ValueError:
        return "Invalid number"
    
    if operator not in ["+", "-", "*", "/", "%", "**"]:
        return "Invalid operator"
    elif operator in ["/", "%"] and b == 0:
        return "Cannot divide by zero"
    else:
        match operator:
            case "+":
                return round(a + b, 2)
            case "-":
                return round(a - b, 2)
            case "*":
                return round(a * b, 2)
            case "/":
                return round(a / b, 2)
            case "%":
                return round(a % b, 2)
            case "**":
                return round(a ** b, 2)
```

```python
def score_summary(name, a, b, c):
    try:
        a = int(a)
        b = int(b)
        c = int(c)
    except:
        return "Invalid score"
    
    grade = ""

    if min(a, b, c) < 0 or max(a, b, c) > 100:
        return "Invalid score"
    else:
        total = a + b + c
        average = round(total / 3, 2)
        if average < 70: grade = "F"
        elif average < 80: grade = "C"
        elif average < 90: grade = "B"
        else: grade = "A"

    return f"Student: {name}\nAverage: {average}\nGrade: {grade}"
```

```python
def sum_multiples(limit, divisor):
    if divisor == 0:
        return "Invalid divisor"
    else:
        total = 0
        for i in range(1, limit + 1):
            if i % divisor == 0:
                total = total + i
        
        return total
```

```python
def skip_evens(start, end):
    if start > end:
        return []
    else:
        result = []
        for i in range(start, end + 1):
            if i % 2 == 0:
                continue
            else:
                result.append(i)
        
        return result
```

```python
def triangle_pattern(n):
    result = ""
    if n < 1:
        return ""
    else:
        for i in range(1, n + 1):
            asterisks = "*"
            if i != n:
                result += (asterisks * i) + "\n"
            else:
                result += (asterisks * i)
    
    return result
```

```python
def prime_list(n):
    result = [2]
    if n < 2:
        return []
    elif n == 2:
        return result

    for i in range(3, n + 1):
        is_prime = True
        for j in range(2, i):
            if i % j == 0:
                is_prime = False
                break
            else:
                continue
        if is_prime == True:
            result.append(i)
    
    return result

prime_list(5)
```

```python
def chunk_list(items, size):
    if size < 1:
        return "Invalid size"
    else:
        result = []
        list_bits = []
        for i in range(len(items)):
            list_bits.append(items[i])
            if len(list_bits) == size:
                result.append(list_bits)
                list_bits = []
            
        if list_bits:
            result.append(list_bits)
    
    return result

chunk_list([1,2,3,4,5], 2)
```

```python
# Another method in list is extend, and extend allows you to expand the items in a list so that you can apply
# recursion to it and append the results into a flattened manner.
def deep_flatten(items):
    result = []
    for item in items:
        if isinstance(item, list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)

    return result

deep_flatten([1,[[[[2,3]]]],4])
```

```python
def top_scorer(scores):
    if len(scores) == 0:
        return "No scores"
    
    top_student = None
    highest_score = None

    for name, score in scores.items():
        if top_student == None and highest_score == None:
            top_student = name
            highest_score = score
            continue
        
        if score > highest_score:
            highest_score = score
            top_student = name
        elif score == highest_score:
            if name < top_student:
                top_student = name
    
    return top_student

# OR

def top_scorer(scores):
    if len(scores) == 0:
        return "No scores"
    
    top_name = ""
    best_score = 0

    for name, score in scores.items():
        top_name = name
        best_score = score

    for name, score in scores.items():
        if score > best_score:
            top_name = name
            best_score = score
            continue

        if score == best_score and name < top_name:
            top_name = name
            
    return top_name


top_scorer({"Ada":90,"Aunde":95,"Ablima":95, "test": 20})
```

```python
def decrypt_brief(message, shift):
    result = ""
    for letter in message:
        if letter.islower():
            shifted_letter = chr(((ord(letter) - ord("a")) - shift) % 26 + ord("a"))
            result += shifted_letter
        elif letter.isupper():
            shifted_letter = chr(((ord(letter) - ord("A")) - shift) % 26 + ord("A"))
            result += shifted_letter
        else:
            result += letter

    return result
```

```python
def longest_streak(log):
    current_streak = 1
    max_streak = 0
    if log == []:
        return 0
    elif len(log) == 1:
        return 1

    for i in range(1, len(log)):
        if log[i] == log[i-1]:
            current_streak += 1
        else:
            if current_streak > max_streak:
                max_streak = current_streak
            
            current_streak = 1
    
    if current_streak > max_streak:
        max_streak = current_streak
    
    return max_streak
```
