```python
# Attempt 2
def simp_calc():
    results = [0]
    operators = ["+","-", "*", "/", "%", "**","//"]
    
    while True:
        a = input("Start the calculator with any operator and a number, e.g: '+ 5' ")

        if a.lower() == "undo" and len(results) > 1:
            return results[-2]
        elif a.lower() == "undo" and len(results) == 1:
            return results[0]
        
        a = a.split()
    
        if a[0] not in operators:
            return "Invalid Operator"

        op = a[0]
    
        try:
            number = float(a[1])
        except:
            return "Please Input a valid number"
        
        if (op == "/" or op == "//") and number == 0:
            return "Cannot divide by zero"

        if op == "+":
            answer = results[-1] + number
            print(answer)
            results.append(answer)
        elif op == "-":
            answer = results[-1] - number
            print(answer)
            results.append(answer)
        elif op == "*":
            answer = results[-1] * number
            print(answer)
            results.append(answer)
        elif op == "/":
            answer = results[-1] / number
            print(answer)
            results.append(answer)
        elif op == "**":
            answer = results[-1] ** number
            print(answer)
            results.append(answer)
        elif op == "%":
            answer = results[-1] % number
            print(answer)
            results.append(answer)
        elif op == "//":
            answer = results[-1] // number
            print(answer)
            results.append(answer)

simp_calc()
```

```python
def student_grades(students):
    highest_grade = 0
    highest_student = ""
    lowest_grade = 101
    lowest_student = ""
    
    for student in students:
        name = student["name"]

        score_a = student["scores"][0]
        score_b = student["scores"][1]

        average = round((score_a + score_b)/ 2, 2)

        if average > highest_grade:
            highest_grade = average
            highest_student = name
        elif average < lowest_grade:
            lowest_grade = average
            lowest_student = name
            
        if average > 70:
            grade = "A"
            print(f"{name} → Average: {average} → Grade: {grade}")
        elif average > 60:
            grade = "B"
            print(f"{name} → Average: {average} → Grade: {grade}")
        elif average > 50:
            grade = "C"
            print(f"{name} → Average: {average} → Grade: {grade}")
        elif average > 45:
            grade = "D"
            print(f"{name} → Average: {average} → Grade: {grade}")
        elif average > 40:
            grade = "E"
            print(f"{name} → Average: {average} → Grade: {grade}")
        else:
            grade = "F"
            print(f"{name} → Average: {average} → Grade: {grade}")
            
    print(f"Highest Average = {highest_grade}, Highest Student = {highest_student}")
    print(f"Lowest Average = {lowest_grade}, Lowest Student = {lowest_student}")
    return

students = [{"name": "Sam", "scores": [80, 90]}, {"name": "Drake", "scores": [75, 50]}, {"name": "David", "scores": [55, 60]}]
student_grades(students)
```

```python
def second_largest(numbers):
    if len(numbers) == 0:
        return "Put a valid list"
    
    largest = numbers[0]
    second = numbers[0]

    for num in numbers:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num < largest:
            second = num

    print(second)
    return

numbers = [10, 5, 18, 20, 15]
second_largest(numbers)

numbers = [4, 9, 12, 9, 7]
second_largest(numbers)
```

```python
def word_frequency(sentence):
    result = {}
    most = ""
    most_num = 1
    new = []

    sentence = sentence.lower()
    sentence = sentence.split()

    for words in sentence:
        text = ""
        for ch in words:
            if ch.isalnum():
                text += ch
        new.append(text)

    for word in new:
        if word not in result:
            # count = 1
            # result[word] = count
            result[word] = 1
        else:
            # result[word] = count + 1
            result[word] += 1

    for word, value in result.items():
        if value > most_num:
            most = word
            most_num = value

    print(f"Most frequent word: {most} ({most_num})")
    # print(result)
    return

sentence = "Python is fun, and  python Python is powerful!"
word_frequency(sentence)
```

```python
def two_sum(numbers, target):
    result = []
    
    for i in range(len(numbers)):
        holder = target - numbers[i]
        if holder in numbers:
            result.append(i)

    return result

numbers = [2, 7, 11, 15]
target = 9
two_sum(numbers, target)

numbers = [5, 7, 11, 5]
target = 10
two_sum(numbers, target)
```

```python
def anagram_finder(word1, word2):
    word1 = list(word1.lower())
    word2 = list(word2.lower())
    copy = []

    for letters in word1:
        copy.append(letters)

    for i in range(len(word1)):
        if word1[i] in word2:
            word2.remove(word1[i])
            copy.remove(word1[i])
            
            
    if copy == [] and word2 == []:
        print("Anagram")
    else:
        print("Not an anagram")

    return

anagram_finder("silent", "listen")
anagram_finder("triangle", "integral")
anagram_finder("pale", "apple")
anagram_finder("apple", "pale")
anagram_finder("hello", "billion")
```

```python
def caesar_cipher():
    while True:
        word = input("What would you like to work on? ")
        if word.lower() == "quit":
            print("Have a nice day.")
            break
            
        shift = input("What would you like to shift it by? ")
        try:
            shift = int(shift)
        except:
            print("Please input a number.")
            continue
            
        mode = input("Would you like to 'Encrypt' or 'Decrypt' the word? ").lower()

        if mode != "encrypt" or mode != "decrypt":
            print("This program can only perform an encryption or a decryption.")
            continue

        result = ""

        for letter in word:
            if letter.islower() and letter.isalpha() and mode == "encrypt":
                result += chr((((ord(letter) - ord("a")) + shift) % 26) + ord("a"))
            elif letter.isupper() and letter.isalpha() and mode == "encrypt":
                result += chr((((ord(letter) - ord("A")) + shift) % 26) + ord("A"))
            elif letter.isupper() and letter.isalpha() and mode == "decrypt":
                result += chr((((ord(letter) - ord("A")) - shift) % 26) + ord("A"))
            elif letter.islower() and letter.isalpha() and mode == "decrypt":
                result += chr((((ord(letter) - ord("a")) - shift) % 26) + ord("a"))
            else:
                result += letter
        print(result)
        continue

caesar_cipher()
```

```python
def number_reverser(number):
    if number == 0:
        print(0)
        return
        
    negative = False
    if number < 0:
        negative = True
        number = number * -1
        
    result = 0

    while number > 0:
        holder = number % 10
        number = number // 10
        result = (result * 10) + holder

    if negative == True:
        result *= -1
    
    print(result)
    return

number = 12345
number_reverser(number)

number = -987
number_reverser(number)

number = 1200
number_reverser(number)

number = 0
number_reverser(number)
```
