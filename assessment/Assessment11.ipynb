```python
def fizzbuzz_plus(n):
    try:
        n = int(n)
    except:
        return "Invalid data type"
    
    result = []

    for i in range(1, n+1):
        if i % 7 == 0 and i % 5 == 0 and i % 3 == 0:
            result.append("FizzBuzzBazz")
        if i % 7 == 0 and i % 5 == 0 and i % 3 != 0:
            result.append("BuzzBazz")
        if i % 7 == 0 and i % 3 == 0 and i % 5 != 0:
            result.append("FizzBazz")
        if i % 5 == 0 and i % 3 == 0 and i % 7 != 0:
            result.append("FizzBuzz")
        if i % 7 == 0 and i % 3 != 0 and i % 5 != 0:
            result.append("Bazz")
        if i % 5 == 0 and i % 3 != 0 and i % 7 != 0:
            result.append("Buzz")
        if i % 3 == 0 and i % 5 != 0 and i % 7 != 0:
            result.append("Fizz")
        if i % 3 != 0 and i % 5 != 0 and i % 7 != 0:
            result.append(str(i))

        
    return result

# Alternative Solution
def fizzbuzz_plus_redo(n):
    try:
        n = int(n)
    except:
        return "Invalid data type"
    
    result = []
    holder = {3: "Fizz", 5: "Buzz", 7: "Bazz"}

    for x in range(1, n + 1):
        result.append("".join(word for num, word in holder.items() if x % num == 0) or str(x))
        
    return result

print(fizzbuzz_plus(25))
print(fizzbuzz_plus_redo(25))
```

```python
def stats(nums):
    if len(nums) == 0:
        return None
        
    low = float("inf")
    high = float("-inf")
    total = 0
    for i in nums:
        if i < low:
            low = i
        if i > high:
            high = i
        total += i
    average = round(total/len(nums), 2)
    return (low, high, average)

print(stats([4, 9, 2, 7]))
print(stats([5]))
print(stats([-3, -1, -2]))
print(stats([]))
```

```python
def count_letters(text):
    text = text.lower()
    vowels = ["a", "e", "i", "o", "u"]
    vowel_count = 0
    consonant_count = 0
    for letter in text:
        if ord(letter) >= ord("a") and ord(letter) <= ord("z"):
            if letter in vowels:
                vowel_count += 1
                continue
            else:
                consonant_count += 1
                continue
    return (vowel_count, consonant_count)

print(count_letters("Hello, World 123"))
print(count_letters("AEIOU"))
print(count_letters("rhythm"))
print(count_letters(""))
```
