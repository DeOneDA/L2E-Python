```python
def fibonacci_sequence(n):
    index = 2
    second = 0
    last = 1
    total_even = 0
    sequence = "0, 1,"
    while index < n:
        holder = last + second
        second = last
        last = holder
        index += 1
        if last % 2 == 0:
            total_even += last
        if index == n:
            sequence += " " + str(last)
            continue
        sequence += " " + str(last) + ","
        continue

    reverse = ""
    holder = ""
    for ch in range(len(sequence)-1, -1, -1):
        if sequence[ch] != " " and sequence[ch] != ",":
            holder = sequence[ch] + holder
            continue
            
        if sequence[ch] == " " and ch != 0:
            reverse += holder + ", "
            holder = ""
            continue

        if sequence[ch] != ",":
            holder += sequence[ch]

    reverse += holder
    
    print(f"Sequence: {sequence}")
    print(f"Even Sum: {total_even}")
    print(f"Reversed: {reverse}")
    return
    
    

fibonacci_sequence(5)
```

```python
def decimal_to_binary(n):
    holder = []
    result = ""
    while n > 0:
        remainder = n % 2
        holder.append(remainder)
        n = n // 2
        continue
        
    for i in range(len(holder)-1, -1, -1):
        result += str(holder[i])

    return result


decimal_to_binary(25)
        
```

```python
def remove_spaces_and_capitalize_words(text):
    text = text.strip()
    text_list = text.split()

    result = ""

    for word in text_list:
        if word[0].islower():
            # result += word[0].upper() + word[1:]
            result += chr(ord(word[0]) - 32) + word[1:]
        else:
            result += word
    
    return result

print(remove_spaces_and_capitalize_words("hello world"))
print(remove_spaces_and_capitalize_words("hello World"))
print(remove_spaces_and_capitalize_words("hello world from python"))
```

```python
def anagram_checker(a, b):
    a = a.lower()
    b = b.lower()

    dict1 = {}
    dict2 = {}

    count = 1
    for ch in a:
        if ch in dict1:
            dict1[ch] += 1
            continue
        dict1[ch] = count

    for ch in b:
        if ch in dict2:
            dict2[ch] += 1
            continue
        dict2[ch] = count

    for letter, count in dict1.items():
        if letter not in dict2:
            return "Not an anagram"
        if dict2[letter] != count:
            return "Not an anagram"

    return "Anagram"

a = "Dormitory"
b = "dirtyroom"

anagram_checker(a,b)
```

```python
def roman_to_integer(text):
    romans = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000
    }

    index = 0
    result = 0
    while index < len(text):
        if index < len(text)-1:
            if romans[text[index+1]] > romans[text[index]]:
                result += romans[text[index+1]] - romans[text[index]]
                index += 2
                continue
        result += romans[text[index]]
        index += 1
        continue

    return result

text = "MCMXCIV"
print(roman_to_integer(text))

text1 = "LVIII"
print(roman_to_integer(text1))
```
