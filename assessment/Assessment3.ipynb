```python
def palindrome(sentence):
    sentence = sentence.replace(" ", "")
    sentence = sentence.lower()
    j = -1
    for i in (sentence):
        if i.isalnum() and sentence[j].isalnum():
            if i != sentence[j]:
                return "Not a palindrome"
            else:
                j = j - 1
                continue
        
        if not i.isalnum():
            continue
        
        if not sentence[j].isalnum():
            j = j - 1
            if i != sentence[j]:
                return "Not a palindrome"
            else:
                j = j - 1
                continue  

    return "Palindrome"

user = input("Test Palindrome? ")
palindrome(user)
```

```python
def student_grades(students):
    highest_average = 0
    highest_student = ""
    lowest_average = float("inf")
    lowest_student = ""
    
    for student in students:
        name = student["name"]
        a, b = student["scores"]
        average = (a + b)/2

        if highest_average == 0 and lowest_average == 0:
            highest_average = average
            lowest_average = average
            highest_student = name
            lowest_student = name
        elif average < lowest_average:
            lowest_average = average
            lowest_student = name
        elif average > highest_average:
            highest_average = average
            highest_student = name
            
        if average >= 70: grade = "A"
        elif average >= 60: grade = "B"
        elif average >= 50: grade = "C"
        elif average >= 45: grade = "D"
        elif average >= 40: grade = "E"
        else: grade = "F"
        print(f"{name} → Average: {average} → Grade: {grade}")
    
    print(f"Highest Average -> {highest_average}, Student Name -> {highest_student}")
    print(f"Lowest Average -> {lowest_average}, Student Name -> {lowest_student}")
    return

students = [{"name": "Sam", "scores": [80, 90]}, {"name": "David", "scores": [55, 60]}, {"name": "Javis", "scores": [65, 70]}]

student_grades(students)

# student = {"scores": [80,90]}
# a, b = student["scores"]
# print(a)
# print(b)
```

```python
def second_largest(numbers):
    second = 0
    largest = 0
    for i in range(0, len(numbers)):
        if second == 0 and largest == 0:
            second = numbers[i]
            largest = numbers[i]
        elif numbers[i] > largest:
            second = largest
            largest = numbers[i]
        elif numbers[i] < largest and numbers[i] > second:
            second = numbers[i]
        
    print(f"Output: {second}")
    print(largest)
    return

numbers = [10, 5, 18, 20, 15]
second_largest(numbers)

numbers = [4, 9, 7, 9, 5]
second_largest(numbers)
```

```python
def two_sum(numbers, target):
    res = []
    for i in range(len(numbers)):
        test = target - numbers[i]
        if test in numbers:
            res.append(i)
        else:
            continue

    return f"{res[0]} + {res[1]} = {target}"

two_sum([5, 7, 11, 2], 9) 
```
