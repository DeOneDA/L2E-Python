```python
def list_analyser(item):
    min_num = item[0]
    max_num = item[0]
    total = item[0]
    avg = item[0]
    if len(item) == 1:
        return f"""Smallest: {min_num}
Largest: {max_num}
Total: {total}
Average: {avg}
"""
    else:   
        for n in range(1, len(item)):
            if item[n] < min_num:
                min_num = item[n]
            if item[n] > max_num:
                max_num = item[n]
            total += item[n]
        avg = round(total/len(item), 1)
        return f"""Smallest: {min_num}
Largest: {max_num}
Total: {total}
Average: {avg}
"""

print(list_analyser([12, 5, 27, 8, 14]))
print(list_analyser([14]))
```

```python
def linear_search(items, target):
    if len(items) == 0:
        return "No list"

    try:
        target = int(target)
    except:
        return "Invalid Target"
        
    for i in range(len(items)):
        if items[i] == target:
            return f"Found at index {i}"
        else:
            continue
            
    return "Target not found"

nums = [14, 7, 22, 9, 7, 31]
print(linear_search(nums, 22))
print(linear_search(nums, 18))
```

```python
def matrix_row_sum(items):
    result = {}
    num = 1
    for matrix in items:
        total = 0
        for n in matrix:
            total += n
            continue
        result[num] = total
        num += 1
        
    return result

matrix = [[2, 4, 6], [1, 3, 5], [10, 20, 30]]
res = matrix_row_sum(matrix)
for key, value in res.items():
    print(f"Row {key} total: {value}")
```

```python
def to_do_mini_project():
    tasks = []
    display = {}

    print("Welcome to Flow:")
    print("Options Available: 'Add', 'Remove', 'View', 'Quit'")
    
    while True:
        user_input = input("What would you like to do? ")
        
        if user_input.lower() == "add":
            task = input("What task would you like to add? ")
            tasks.append(task)
            for i in range(len(tasks)):
                display[i+1] = tasks[i]
            print("Task has been added.")
            print("\n", end = "")
            continue
            
        elif user_input.lower() == "remove":
            if len(tasks) == 0:
                print("The task manager is empty, please add a task first.")
                print("\n", end = "")
                continue
                
            task = input("What task would you like to remove(E.g. '1')? ")
            
            try:
                task = int(task)
            except:
                print("Incorrect task number")
                print("\n", end = "")
                continue
                
            display.pop(task)
            
            for i in range(len(tasks)):
                if i == task -1:
                    tasks.remove(tasks[i])
            print("Task has been successfully removed")
            print("\n", end = "")
            continue
            
        elif user_input.lower() == "view":
            if len(tasks) == 0:
                print("The task manager is empty, please add a task first.")
                print("\n", end = "")
                continue
            for key, value in display.items():
                print(f"Task {key}: {value}")
            continue
            
        elif user_input.lower() == "quit":
            print("Have a nice day")
            break
            
        else:
            print("Please put in a valid action")
            print("\n", end = "")
            continue

to_do_mini_project()
```
