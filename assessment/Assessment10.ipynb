```python
def word_break(s, word_dict, index = 0):
    holder = ""
    
    while index < len(s):
        if holder in word_dict:
            s = s[index:]
            holder = ""
            index += 1
            return word_break(s, word_dict, index)
        if holder not in word_dict:
            holder += s[index]
            index += 1
            return word_break(s, word_dict, index)

    if s == "":
        return True
        
    if s != "":
        return False

word_break("leetcode", ["leet", "code"])
# word_break("applepenapple", ["apple", "pen"])
```

```python
def is_interleave(s1, s2, s3, i=0, j=0):
    k = i + j
    if i == len(s1) and j == len(s2) and k == len(s3):
        return True

    if i < len(s1) and j < len(s2):
        if s1[i] == s3[k] and s2[j] == s3[k]:
            return is_interleave(s1, s2, s3, i+1, j) or is_interleave(s1, s2, s3, i, j+1)
    if i < len(s1):
        if s1[i] == s3[k]:
            i += 1
            return is_interleave(s1, s2, s3, i, j)
    if j < len(s2):
        if s2[j] == s3[k]:
            j += 1
            return is_interleave(s1, s2, s3, i, j)

    return False

# s1 = "abc"
# s2 = "axf"
# s3 = "aabcxg"
# is_interleave(s1, s2, s3)
# s1 = "abc"
# s2 = "def"
# s3 = "abdfec"
# is_interleave(s1, s2, s3)
s1 = "aab"
s2 = "axy"
s3 = "aaxaby"
is_interleave(s1, s2, s3)
```

```python
def evaluate(expr, index = 0):
    num = 0
    result = 0
    try:
        if int(expr[index]):
            num *= 10
            num += int(expr[index])
            index += 1
            evaluate(expr, index)
    except:
        if index < len(expr) and expr[index] == "*":
            index += 1
            result += num * evaluate(expr, index)
        elif index < len(expr) and expr[index] == "+":
            index += 1
            result += num + evaluate(expr, index)
        elif index < len(expr) and expr[index] == "-":
            index += 1
            result += num - evaluate(expr, index)

    return result

print(evaluate("8*2+6"))
```

```python
def josephus(n, k):
    if n == 1:
        return 0

    return (josephus(n-1, k) + k) % n

josephus(7, 3)
```
