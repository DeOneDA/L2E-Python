```python
def binary_search(nums, target):
    if len(nums) == 0:
        return (-1, 0)

    comparisons = 0
    i = 0
    left = 0
    while i < len(nums):
        if nums[i] == target:
            left = i
            comparisons += 1
            break
        comparisons += 1
        i += 1
        continue

    if left == 0:
        return (-1, comparisons)

    return (left, comparisons)

nums1 = [1, 3, 3, 3, 5, 8]
target1 = 5

print(binary_search(nums1, target1))

nums2 = [1, 3, 5, 8]
target2 = 9

print(binary_search(nums2, target2))
```

```python
def bubble_sort(items):
    count = 0

    for i in range(len(items)):
        count += 1
        switch = True
        for j in range(i+1, len(items)):
            if items[j] < items[i]:
                holder = items[i]
                items[i] = items[j]
                items[j] = holder
                switch = False
        if switch:
            break

    return items, count

print(bubble_sort([5, 1, 4, 2, 8]))
print(bubble_sort([1, 2, 3, 4, 5]))
print(bubble_sort([]))
```

```python
def pascal_triangle(n):
    if n == 0:
        return []
    if n == 1:
        return [1]
    if n == 2:
        return [1, 1]

    pascal = [[1], [1, 1]]

    holder = [1]
    last = pascal[-1]
    index = 0
    
    for i in range(3, n+1):
        while len(holder) < i-1:
            holder.append(last[index] + last[index+1])
            index += 1
        holder.append(1)
        pascal.append(holder)
        holder = [1]
        index = 0
        last = pascal[-1]

    return pascal

pascal_triangle(10)
```
