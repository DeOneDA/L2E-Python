```python
def list_analyzer(sample):
    largest = sample[0]
    largest_index = sample[0]
    smallest = sample[0]
    smallest_index = sample[0]
    avg = 0
    total = 0
    num_above_avg = 0

    for num in range(len(sample)):
        if sample[num] > largest:
            largest_index = num
            largest = sample[num]
        if sample[num] < smallest:
            smallest_index = num
            smallest = sample[num]
        total += sample[num]

    avg = round(total / len(sample), 2)

    for n in sample:
        if n > avg:
            num_above_avg += 1
        else:
            continue

    return f"""Smallest: {smallest} (at index {smallest_index})
Largest: {largest} (at index {largest_index})
Total: {total}
Average: {avg}
Numbers above average: {num_above_avg}"""

numbers = [12, 5, 27, 8, 14, 27, 3]
print(list_analyzer(numbers))

print("~" * 10)

number2 = [0, 35, 20, 6, 100, 5, 13]
print(list_analyzer(number2))
```

```python
def linear_search(values, target):
    if target not in values:
        print("Target not found")
        return

    indices = []
    count = 0

    for i in range(len(values)):
        if values[i] == target:
            indices.append(i)
            count += 1
        else:
            continue

    first = indices[0]
    last = indices[-1]

    return f"Found at indices: {indices} | Count: {count} | First: {first} | Last: {last}"

values = [14, 7, 22, 9, 7, 31, 7, 5]
target = 7

print(linear_search(values, target))

print("~" * 10)

values_1 = [14, 7, 22, 9, 7, 31, 7, 5]
target_1 = 0

print(linear_search(values_1, target_1))
```

```python
def matrix_row_sums(matrix):
    row_totals = len(matrix) * [0]
    row_avgs = len(matrix) * [0]
    length = len(matrix[0])
    col_totals = len(matrix[0]) * [0]
    
    for row in range(len(matrix)):
        total = 0
        for i in matrix[row]:
            total += i
        row_totals[row] = total

    for rows in range(len(matrix)):
        test = matrix[rows]
        for j in range(len(test)):
            col_totals[j] += test[j]

    for k in range(len(row_totals)):
        row_avgs[k] = round(row_totals[k] / length, 2)

    for a in range(len(row_totals)):
        print(f"Row {a+1} total: {row_totals[a]} | Row {a+1} average: {row_avgs[a]}")

    print()

    for b in range(len(col_totals)):
        print(f"Column {b+1} total: {col_totals[b]}")

    return


matrix = [[2, 4, 6], [1, 3, 5], [10, 20, 30]]
matrix_row_sums(matrix)

print("~" * 20)

matrix_2 = [[2, 4, 6, 8], [1, 3, 5, 7], [10, 20, 30, 40]]
matrix_row_sums(matrix_2)
```

```python
def balanced_parentheses(text):
    openings = ["(", "[", "{"]
    closings = [")", "]", "}"]

    brackets = ""

    for i in text:
        if i in openings or i in closings:
            brackets += i
        else:
            continue

    tester = []

    for a in range(len(brackets)):
        if brackets[0] not in openings:
            return "Not balanced (wrong order)"

        if brackets[a] in openings:
            tester.append(brackets[a])
        else:
            if tester:
                last = tester[-1]
                index = openings.index(last)
                closer = closings[index]
                if brackets[a] == closer:
                    tester.pop()
                else:
                    return "Not balanced (wrong order)"
            elif len(tester) == 0:
                return "Not balanced (mismatched type)"

    if tester:
        return "Not balanced (unclosed)"


    return "Balanced"
    

        

test = "(a + b) * [c - d]"
test_1 = "((a + b)"
test_2 = "(a + b)]" 
test_3 = "([)]"
test_4 = "}}[)]"

print(balanced_parentheses(test))
print(balanced_parentheses(test_1))
print(balanced_parentheses(test_2))
print(balanced_parentheses(test_3))
print(balanced_parentheses(test_4))
```

```python
def to_do_mini_project():
    print("To do Project")
    storage = []
    while True:
        print("Menu: add | view | remove | done | quit")
        user = input("what would you like to do? ").lower().strip()
        menu = ["add", "view", "remove", "done", "quit"]
        if user not in menu:
            print("Invalid input, please type an option in the menu")
            print()
            continue

        if user == "add":
            task = input("What would you like to add? ")
            storage.append([task, False])
            print("Task added.")
            print()
            
        elif user == "view":
            if len(storage) > 0:
                for i in range(len(storage)):
                        print(f"{i+1}. [{"X" if storage[i][1] else " "}] {storage[i][0]}")
                        continue
                print()
                continue
            else:
                print("Nothing has been added to storage")
                print()
                continue
        elif user == "remove":
            if storage:
                num = input("What would you like to remove? ")
                try:
                    num = int(num)
                except:
                    print("Please put the task number")
                    print()
                    continue
                if num > len(storage):
                    print(f"You only have {len(storage)} tasks in storage.")
                    print()
                    continue
                else:
                    storage.pop(num-1)
                    print("Task removed.")
                    print()
            else:
                print("Nothing has been added to storage")
                print()
                continue
        elif user == "done":
            if storage:
                a = input("What task did you complete? ")
                try:
                    a = int(a)
                except:
                    print("Please put the task number")
                    print()
                    continue
                if a > len(storage):
                    print(f"You only have {len(storage)} tasks in storage.")
                    print()
                    continue
                else:
                    storage[a-1][1] = True
                    print(f"Task {a} marked as done.")
                    print()
                    continue
            else:
                print("Nothing has been added to storage")
                print()
                continue
        elif user == "quit":
            print("Goodbye!")
            return


to_do_mini_project()
```

```python
def leap_year_checker():
    year = input("What year would you like to check? ")
    try:
        year = int(year)
    except:
        return "Invalid year"

    if year % 100 != 0:
        if year % 4 == 0:
            return f"{year} is a leap year"
    elif year % 100 == 0 and year % 400 == 0:
        return f"{year} is a leap year"
    else:
        return f"{year} is not a leap year"

    return f"{year} is not a leap year"

leap_year_checker()
```

```python
def find_duplicates(array):
    res = []

    for i in range(len(array)):
        count = 1
        for j in range(i+1, len(array)):
            if array[j] == array[i]:
                count += 1
        if count > 1 and array[i] not in res:
            res.append(array[i])


    return res

test = [1, 2, 3, 2, 4, 5, 1]
print(find_duplicates(test))

test1 = [1, 2, 3, 4]
print(find_duplicates(test1))

test2 = [4, 4, 4, 4]
print(find_duplicates(test2))

test3 = [10, 20, 10, 30, 20]
print(find_duplicates(test3))
```

```python
def run_length_encoding(text):
    res = ""
    text = text.strip()
    sample = []
    for ch in text:
        sample.append(ch)

    res = []
     # "aaabbaa"
    for item in sample:
        if len(res) == 0:
            res.append(item)
            res.append(1)
        elif item == res[-2]:
            res[-1] += 1
        elif res[-2] != item:
            res.append(item)
            res.append(1)

        
    result = ""
    
    for item in res:
        if type(item) == str:
            result += item
        else:
            result += str(item)

    return result
    

test = "aaabbc"
print(run_length_encoding(test))

test1 = "abcd"
print(run_length_encoding(test1))

test2 = "aaa"
print(run_length_encoding(test2))

test3 = "aabbaa"
print(run_length_encoding(test3))


# def decoder(text):
#     res = ""
#     for i in range(0, len(text), 2):
#         res += text[i] * int(text[i+1])

#     return res

# run_length_encoding(test)
# # decoder(tester)
```
