```python
def lru_cache(max_size):
    cache = {}
    def decorator(func):
        def wrapper(*args):
            wrapper.calls += 1
            nonlocal max_size
            res = func(*args)
            if max_size == None or max_size > 0:
                if max_size == None:
                    max_size = float('inf')
                if len(cache) < max_size:
                    if args not in cache:
                        wrapper.misses += 1
                        cache[args] = res
                    elif args in cache:
                        wrapper.hits += 1
                        cache.pop(args)
                        cache[args] = res
                else:
                    if args not in cache:
                        wrapper.misses += 1
                        for key, value in cache.items():
                            cache.pop(key)
                            break
                        cache[args] = res
                    elif args in cache:
                        wrapper.hits += 1
                        cache.pop(args)
                        cache[args] = res
            def get_keys():
                res = list(cache.keys())
                return res

            def clear():
                wrapper.hits = 0
                wrapper.misses = 0
                wrapper.calls = 0
                cache.clear()
                return

            wrapper.keys = get_keys
            wrapper.clear = clear
            return res
        wrapper.cache = cache
        wrapper.hits = 0
        wrapper.misses = 0
        wrapper.calls = 0
        return wrapper
    return decorator


def fifo(max_size):
    cached = {}
    def decorator(func):
        def wrapper(*args):
            nonlocal max_size
            wrapper.calls += 1
            res = func(*args)
            if max_size == None or max_size > 0:
                if max_size == None:
                    max_size = float('inf')
                if len(cached) < max_size:
                    if args not in cached:
                        wrapper.misses += 1
                        cached[args] = res
                    elif args in cached:
                        wrapper.hits += 1
                else:
                    if args not in cached:
                        wrapper.misses += 1
                        for key, value in cached.items():
                            cached.pop(key)
                            break
                        cached[args] = res
                    elif args in cached:
                        wrapper.hits += 1

            def get_keys():
                res = list(cached.keys())
                return res

            def clear():
                wrapper.hits = 0
                wrapper.misses = 0
                wrapper.calls = 0
                cached.clear()
                return

            wrapper.keys = get_keys
            wrapper.clear = clear
            return res
        wrapper.cache = cached
        wrapper.hits = 0
        wrapper.misses = 0
        wrapper.calls = 0
        return wrapper
    return decorator

@lru_cache(max_size=2)
def multiply(x):
    return x*x


# multiply(1)
# multiply(1)
# multiply(1)
# multiply(3)
# multiply(5)
# multiply(5)
# multiply(3)
# multiply(7)
# print(multiply.cache)
# print(multiply.keys)
# print(multiply.misses)
# print(multiply.hits)
# print(multiply.calls)
# print(multiply.keys())
# print(multiply.clear())
# print(multiply.misses)
# print(multiply.hits)
# print(multiply.calls)
# print(multiply.keys())

@fifo(max_size=2)
def add(x):
    return x*x

add(2)
add(2)
add(1)
add(3)
add(1)
add(5)
print(add.cache)
# print(add.keys)

# print("~" * 5)

# @lru_cache(max_size=None)
# def multiply(x):
#     return x*x

# multiply(2)
# multiply(3)
# print(multiply.keys)

# @fifo(max_size=None)
# def add(x):
#     return x*x

# add(2)
# print(add.keys)
```

```python
def count_ways_dp(amount, coins):
    if amount == 0:
        return 1
        
    holder = [1] + [0] * amount
    
    for i in coins:
        for j in range(i, len(holder)):
            holder[j] += holder[j-i]
    
    return holder[-1]

count_ways_dp(5, [1,2,5])
count_ways_dp(100, [1, 5, 10, 25, 50])

def min_coins(amount, coins):
    holder = [0] + [float('inf')] * amount

    for i in coins:
        for j in range(i, len(holder)):
            test = holder[j-i] + 1
            if holder[j] > test:
                holder[j] = test

    return holder[-1]

min_coins(11, [1, 2, 5])
    
```
