```python
def remove_spaces_and_capitalize_words(text):
    capitalized = ""
    for ch in range(len(text)):
        if text[ch].islower() and ch == 0:
            capitalized += chr(ord(text[ch]) - 32)
            continue
        if text[ch-1] == " " and text[ch].islower():
            capitalized += chr(ord(text[ch]) - 32)
            continue

        capitalized += text[ch]


    no_spaces = ""
    word_list = capitalized.split()

    for word in word_list:
        no_spaces += word

    return f'''Capitalized: "{capitalized}"
No Spaces: "{no_spaces}"'''


test = "Hello World From Python"
test1 = "hello world from python"
print(remove_spaces_and_capitalize_words(test))
print()
print(remove_spaces_and_capitalize_words(test1))
```

```python
def prime(n):
    index = 2
    while index <= n**0.5:
        if n % index == 0:
            return False
        index += 1
    return True

def prime_number_checker(func, n):
    if n == 2:
        return f""" {2} is a prime number
Primes up to 2: {[2]}
Sum of primes: {2}"""
    
    index = 3
    primes = [2]

    while index <= n:
        switch = func(index)
        if switch:
            primes.append(index)

        index += 1


    total = 0

    for prime in primes:
        total += prime

    if func(n):
        return f"""{n} is a prime.
Primes up to {n}: {primes}
Sum of primes: {total}"""

    return f"""{n} is not prime.
Primes up to {n}: {primes}
Sum of primes: {total}"""


print(prime_number_checker(prime, 10))
```

```python
def matrix_multiplication(a, b):
    rows = len(a)
    cols = len(b[0])
    cols_A = len(a[0])
    result = [[0 for _ in range(rows)] for _ in range(cols)]

    for i in range(rows):
        for j in range(cols):
            for k in range(cols_A):
                result[i][j] += a[i][k] * b[k][j]

    return result

a =  [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]

matrix_multiplication(a, b)
```
