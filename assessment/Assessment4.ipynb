```python
def count_words(words):
    result = {}

    for word in words:
        if word not in result:
            result[word] = 1
        else:
            result[word] += 1

    return result

def most_frequent(freq):
    most = ""
    value = 1
    for word, count in freq.items():
        if count > value:
            most = word
            value = count

    return f"Most frequent word: {most} ({value})"

sentence = input("Test: ").lower()
words = []
text = ""
new = sentence.split()

for word in new:
    for letter in word:
        if letter.isalpha():
            text += letter
    words.append(text)
    text = ""
    continue

freq = count_words(words)
print(most_frequent(freq))
```

```python
def two_sums(nums, target):
    result = []

    if len(nums) == 0:
        return []

    for i in range(len(nums)):
        holder = target - nums[i]

        if holder in nums:
            result.append(i)

    return result

numbers = [1, 5, 7,7,2,6]
target = 9

two_sums(numbers, target)
```

```python
def second_largest(nums):
    largest = nums[0]
    second = nums[0]
    for x in nums:
        if x > largest:
            second = largest
            largest = x
        elif x < largest and x > second:
            second = x

    return second

nums = [4, 9, 2, 9, 7]
second_largest(nums)
```
