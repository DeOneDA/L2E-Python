```python
def grade_for(avg):
    grade = ""
    if avg > 70:
        grade = "A"
    elif avg > 60:
        grade = "B"
    elif avg > 50:
        grade = "C"
    elif avg > 45:
        grade = "D"
    elif avg > 40:
        grade = "E"
    else:
        grade = "F"

    return grade

def student_info(student):
    name = student["name"]
    scores = student["scores"]
    avg = round((scores[0] + scores[1])/2, 2)
    grade = grade_for(avg)
    student_tuple = (name, avg, grade)
    return student_tuple

def pass_students(student):
    grades = ["A", "B", "C"]
    name, avg, grade = student
    if grade in grades:
        return name

students = [{"name": "Sam", "scores": [80, 90]}, {"name": "David", "scores": [55, 60]}, {"name": "Peter", "scores": [55, 40]}]

info = list(map(student_info, students))
print(info)
passed_students = list(filter(pass_students, info))
print(passed_students)

highest_student = ""
highest = 0
lowest_student = ""
lowest = 100

for student in info:
    name, avg, grade = student
    if avg > highest:
        highest_student = name
        highest = avg

    if avg < lowest:
        lowest_student = name
        lowest = avg

print(f"Highest Student: {highest_student} ({highest})")
print(f"Lowest Student: {lowest_student} ({lowest})")
```

```python
# Monday Live Solution
# def second_largest(nums):
#     largest = nums[0]
#     second = nums[0]
#     for x in nums:
#         if x > largest:
#             second = largest
#             largest = x
#         elif x < largest and x > second:
#             second = x
#     return second

# Alternative Solution

def step(largest, second_largest, x):
    a = largest
    b = second_largest
    
    if x > a:
        b = a
        a = x
    elif x > b and x < a:
        b = x

    return a, b

def second_largest(nums):
    if len(nums) == 0:
        return 0

    largest = nums[0]
    second = nums[0]
    
    for i in nums:
        largest, second = step(largest, second, i)

    return second
        
nums = [-4, -9, -2, -9, -7, -10]
print(second_largest(nums))

nums = [10, 5, 8, 20, 15]
print(second_largest(nums))
```

```python
def make_shifter(shift):
    def shifter(letter):
        if letter.isalpha() and letter.islower():
            new = chr(((ord(letter) - ord("a")) + shift) % 26 + ord("a"))
            return new
        elif letter.isalpha() and letter.isupper():
            new = chr(((ord(letter) - ord("A")) + shift) % 26 + ord("A"))
            return new
        else:
            return letter
    return shifter

while True:
    text = input("Caesar Cipher Code: ")
    value = input("Shift Value: ")
    try:
        value = int(value)
    except ValueError:
        print("Wrong value")
        continue
    mode = input("Mode Type: ").lower()

    if mode == "decrypt":
        value *= -1

    shift = make_shifter(value)
    result = list(map(shift, text))
    new_string = ""
    for i in result:
        new_string += i

    print(new_string)
    break
```

```python
def compose(f, g):
    def worker(x):
        start = g(x)
        end = f(start)
        return end
    return worker

f = lambda x: x + 1
g = lambda x: x * 2

tester = compose(f, g)
result = tester(5)
print(result)

def pipe(*funcs):
    def worker(x):
        result = []
        for i in funcs:
            if len(result) == 0:
                answer = i(x)
                result.append(answer)
            else:
                answer = i(result[-1])
                result.append(answer)
        return result[-1]
    return worker

def strip(text):
    return text.strip()

def lower(text):
    return text.lower()

def remove_vowels(text):
    vowels = ["a", "e", "i", "o", "u"]
    new = ""
    for letter in text:
        if letter in vowels:
            continue
        else:
            new += letter
    return new

def reverse(text):
    new = ""
    for i in range(len(text)):
        new = text[i] + new
    return new

# Bonus Constraint
# def reverse(text, index = None):
#     if index == None:
#         index = len(text)-1
#     elif index < 0:
#         return ""
#     return text[index] + reverse(text, index -1)

test = pipe(f, g)
v2 = test(5)
print(v2)

pipeline = pipe(strip, lower, remove_vowels, reverse)
pipeline(" Hello World ")
```
