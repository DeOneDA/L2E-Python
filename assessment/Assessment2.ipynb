```python
def calculator(num1, operator, num2):
    valid = ["+", "-", "*", "/"]
    try:
        operator = str(operator)
    except:
        print("Invalid Operator")
        return

    if operator not in valid:
        print("Invalid Opertor")
        return
        
    try:
        num1 = int(num1)
        num2 = int(num2)
    except:
        print("Put a valid number")
        return

    if operator == "/" and (num1 == 0 or num2 == 0):
        print("Cannot divide by zero")
    elif operator == "+":
        print(num1 + num2)
    elif operator == "-":
        print(num1 - num2)
    elif operator == "*":
        print(num1 * num2)
    else:
        print(round(num1/num2,2))
    
    return

a = input("Enter the first number: ")
b = input("Enter the second number: ")
op = input("Enter the operator: ")

calculator(a, op, b)
```

```python
def palindrome(string):
    try:
        string = str(string)
    except:
        print("Input a valid string")
        return

    length = len(string)-1
    
    for i in range(len(string)):
        if string[i] == string[length]:
            length = length-1
            continue
        else:
            print("String is not a palindrome")
            return

    print("String is a palindrome")

user_input = input("Enter your palindrome string.")
palindrome(user_input)
```

```python
def grades(students):
    if len(students) == 0:
        print("Input a valid list")
        return

    for student in students:
        name = student[0]
        average = round((student[1]+student[2]+student[3])/3, 2)
        if average >= 70:
            grade = "A"
        elif average >= 60:
            grade = "B"
        elif average >= 50:
            grade = "C"
        elif average >= 45:
            grade = "D"
        elif average >= 40:
            grade = "E"
        else:
            grade = "F"
        print(f"{name} -> Average: {average} -> Grade: {grade}")

user_input = [["Samuel", 80, 75, 90], ["David", 55, 60, 50], ["Mary", 35, 40, 30], ["John", 65, 70, 68], ["Damilare", 90, 80, 85]]
grades(user_input)
```

```python
def word_frequency(sentence):
    sentence = sentence.split(" ")
    result = []
    count = 1

    for i in range(len(sentence)):
        if sentence[i] not in result:
            result.append(sentence[i])
        else:
            continue
        if i != len(sentence)-1:
            for j in range(i+1, len(sentence)):
                if sentence[j] == sentence[i]:
                    count += 1
            result.append(count)
            count = 1
        else:
            result.append(count)

    print(result)
    return

word_frequency("python is fun and python is is is powerful")
```
