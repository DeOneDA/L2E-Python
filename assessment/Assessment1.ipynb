```python
def calculator(a, operator, b):
    if operator not in ["+", "-", "*", "/"]:
        return "Invalid Operator"

    try:
        a = int(a)
        b = int(b)
    except:
        return "Invalid Numbers"

    try:
        if operator == "/":
            return a / b
    except:
        return "Cannot divide by 0"

    if operator == "+": return a + b
    elif operator == "-": return a - b
    elif operator == "*": return a * b

a = input("What is the first number: ")
b = input("What is the second number: ")
operator = input("What operation do you wish to perform? ")

calculator(a, operator, b)
```

```python
def palindrome(string):
    try:
        string = str(string)
    except:
        return "Invalid String"

    j = len(string)-1
    
    for i in string:
        if i == string[j]:
            j = j-1
            continue
        else:
            return "Not a palindrome"

    return "String is a palindrome"

string = input("What do you want to check?")
palindrome(string)
```

```python
def grades(students):
    for student in students:
        average = round((student[1] + student[2] + student[3])/3, 2)
        if average > 70: Grade = "A"
        elif average > 60: Grade = "B"
        elif average > 50: Grade = "C"
        elif average > 45: Grade = "D"
        elif average > 40: Grade = "E"
        else: Grade = "F"
        print(f"{student[0]} - Average: {average} - Grade: {Grade}")

students = [
    ["Samuel", 80, 75, 90],
    ["David", 55, 60, 50],
    ["Mary", 35, 40, 30],
    ["John", 65, 70, 68],
    ["Dare", 50, 45, 80]
]

grades(students)
```
