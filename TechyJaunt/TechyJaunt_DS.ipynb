```python
top_product = (
    orders.groupby("product_name")["total_amount"].sum()
    .sort_values(ascending = False)
    .head(10)
        )
top_product.plot(kind = "bar")
plt.title("Top Products by Revenue")
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.show()
#print(top_product)
```

```python
monthly_sales = (
    orders.groupby("order_date")["total_amount"].sum()
)
monthly_sales.plot(kind = "line")
plt.show()
```
