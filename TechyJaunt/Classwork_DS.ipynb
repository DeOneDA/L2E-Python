```python
#Total Revenue 
int(orders["total_amount"].sum())
```

```python
#Average Order Value
orders["total_amount"].mean()
```

```python
#Top 5 Products by Revenue
top_5_products = (
    orders.groupby("product_name")["total_amount"].sum()
    .sort_values(ascending = False)
    .head(5)
)
print(top_5_products)
```

```python
import matplotlib.pyplot as plt
```

```python
# Top 5 Products by Revenue Bar Chart
top_5_products = (
    orders.groupby("product_name")["total_amount"].sum()
    .sort_values(ascending = False)
    .head(5)
)
top_5_products.plot(kind = "bar")
plt.title("Top 5 Products by Revenue")
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.show()
```

```python
# Highest Payment Method
top_payment_method = (
    orders.groupby("payment_method")["quantity"].sum()
    .head(1)
)

print(top_payment_method)
```

```python
# Payment Method Pie Chart
payment_method_chart = orders["payment_method"].value_counts()
payment_method_chart.plot(kind = "pie")

#orders["payment_method"].value_counts().plot(kind = "pie")
plt.show()
```

```python
# Total Order Amount Histogram
orders["total_amount"].plot(kind = "hist")
plt.title("Histogram Chart of the Frequency of Revenue")
plt.xlabel("Amount")
plt.show()
```

```python
#Special Condition.
orders.groupby("payment_method")["quantity"].sum().plot(kind = "bar")

plt.show()
```
