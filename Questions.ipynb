```python
def summarize_expenses(lines):
    result = {
        "totals": {},
        "daily_totals": {},
        "biggest_category": "",
        "errors": []
    }
    for line in lines:
        if "|" not in line:
            result["errors"].append(lines.index(line) + 1)
            continue
        date, category, price = line.lower().split("|")
        date = date.strip()
        category = category.strip()
        try:
            price = round(float(price.strip()), 2)
        except ValueError:
            result["errors"].append(lines.index(line) + 1)
            continue
        if category in result["totals"]:
            result["totals"][category] += price
        else:
            result["totals"][category] = price

        if date in result["daily_totals"]:
            result["daily_totals"][date] += price
        else:
            result["daily_totals"][date] = price
            
    maximum = max(result["totals"].values())
    for key, value in result["totals"].items():
        if value == maximum:
            result["biggest_category"] = key

    return result


summarize_expenses([
    "2024-03-01 | FOOD | 12.50",
    "2024-03-01 | transport | 3.20",
    "2024-03-02 |  food  | 8.75",
    "this is not a valid line",
    "2024-03-02 | fun | abc",
    "2024-03-03 | Food | 20.00",
])
```
